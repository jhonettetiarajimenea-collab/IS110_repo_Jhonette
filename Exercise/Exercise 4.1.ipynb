```python
import pandas as pd
import numpy as np
```

```python
df_gdp = pd.read_excel(
    file_path,
    usecols=[0, 2, 3, 4, 5],  # Exclude the Country Code column
    names=["Country", "Life_Exp", "GDP", "Population", "Inc_Class"],
    na_values=["..", "-", "NA", "N/A", "_"]
)
df_gdp.head()
```

```python
df_gdp["GDP"] = pd.to_numeric(df_gdp["GDP"], errors="coerce")
df_gdp["Life_Exp"] = pd.to_numeric(df_gdp["Life_Exp"], errors="coerce")
df_gdp["Population"] = pd.to_numeric(df_gdp["Population"], errors="coerce"

print("1. Dimensions:", df_gdp.shape

print("2. Valid GDP values:", df_gdp["GDP"].count())

print("3. Average life expectancy:", df_gdp["Life_Exp"].mean())

df_sorted = df_gdp.sort_values(by="Country")
print("4. Sorted DataFrame:")
display(df_sorted.head())

print("5. Top 5 countries with highest GDP:")
display(df_gdp.nlargest(5, "GDP"))

print("6. Top 5 countries with smallest population:")
display(df_gdp.nsmallest(5, "Population"))
```
